```python
from agents import classificator, translator, causal_A
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
async def main(task=None):
    if task is None:
        task = input("Inserisci la tua domanda causale: ")

    print(f"\n🎯 TASK:\n{task}\n")

    print("🔄 1. Traduzione in corso...")
    translation_response = await translator.run(
        f"Create a dictionary following your instruction of the query: {task}"
    )
    translation_a = translation_response.messages[-1].content if hasattr(translation_response, 'messages') else str(translation_response)
    print(f"📝 Dictionary:\n{translation_a}\n")

    print("🔄 2. Classificazione in corso...")
    label_response = await classificator.run(
        f"Classify the following query using your instructions: {task}"
    )
    label_a = label_response.messages[-1].content if hasattr(label_response, 'messages') else str(label_response)
    print(f"🏷️ Label:\n{label_a}\n")

    print("🔄 3. Calcolo causale in corso...")
    prompt = (
        f"Provide the precise answer to the user task ({task}) "
        f"using your TOOLS with the following parameters -> "
        f"label: {label_a}, features: {translation_a}. "
        f"After that give the causal explanation of this answer."
    )
    causal_response = await causal_A.run(prompt)
    causal_answer_a = causal_response.messages[-1].content if hasattr(causal_response, 'messages') else str(causal_response)
    print(f"✅ Risposta finale:\n{causal_answer_a}\n")

    return translation_a, label_a, causal_answer_a
```

```python
# Test con la prima query del ground truth (CP1)
await main("Given that the driving quality is normal, how likely is it that the age of the driver is senior?")
```
